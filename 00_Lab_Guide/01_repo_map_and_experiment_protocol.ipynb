# Repo Map and Experiment Protocol

This notebook is the entry point for the repository.

It does four things:

1. explains how the repo is organized after the lane-based restructure,
2. clarifies what each maturity label means,
3. recommends a reading path for different kinds of readers,
4. defines the experiment discipline that future notebooks should follow.

The current repo is intentionally honest about maturity: most notebooks are still scaffolds, and this notebook explains how to grow them without letting scope drift again.

## Notebook Contract

- **Status:** scaffold
- **Lane:** `00_Lab_Guide`
- **Primary audience:** researchers or students onboarding into the repo
- **Primary deliverable:** a shared mental model for navigation, scope, and experiment quality
- **What this notebook is not:** a PINN tutorial, a benchmark notebook, or a paper reproduction

## How to Use This Repo

The repository is organized into four lanes.

| Lane | Role | How to use it |
|---|---|---|
| `00_Lab_Guide` | Orientation and working norms | Start here before touching the rest of the repo |
| `01_Core_Path` | Narrow backbone of the repo | Follow in order if you want the minimum coherent PINN journey |
| `10_Research_Tracks` | Optional deeper branches | Use after the core path, based on your research question |
| `20_Frontier_Notes` | Literature-first frontier map | Read when you want breadth and paper directions without promising full replication |

Recommended reading paths:

- **New to PINNs:** read this notebook, then finish all notebooks in `01_Core_Path` in order.
- **Already know the basics:** scan this notebook, then jump into the specific research track closest to your current problem.
- **Exploring frontier methods:** use the frontier lane as a reading map, but anchor yourself in the core path first so the comparisons stay meaningful.

````python
from pathlib import Path
import platform
import sys


def find_repo_root(start: Path) -> Path:
    current = start.resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "pyproject.toml").exists():
            return candidate
    raise FileNotFoundError("Could not find repo root containing pyproject.toml")


root = find_repo_root(Path.cwd())

try:
    import torch

    torch_version = torch.__version__
    cuda_available = torch.cuda.is_available()
except Exception as exc:
    torch_version = f"unavailable ({exc.__class__.__name__})"
    cuda_available = False

lane_counts = {}
for lane in ["00_Lab_Guide", "01_Core_Path", "10_Research_Tracks", "20_Frontier_Notes"]:
    lane_counts[lane] = len(list((root / lane).rglob("*.ipynb")))

print(f"Repo root: {root}")
print(f"Python: {sys.version.split()[0]}")
print(f"Platform: {platform.platform()}")
print(f"PyTorch: {torch_version}")
print(f"CUDA available: {cuda_available}")
print()
for lane, count in lane_counts.items():
    print(f"{lane}: {count} notebook(s)")
print(f"Total notebooks: {sum(lane_counts.values())}")
````

## Maturity Model

The lane a notebook belongs to and the maturity of that notebook are different concepts.

| Label | Meaning in this repo | Expected evolution |
|---|---|---|
| `scaffold` | The notebook exists in the right lane, but the body is still minimal | Fill in the notebook without expanding its scope carelessly |
| `core-executable` | A narrow, runnable notebook in the backbone path | Keep it focused and reproducible |
| `research-executable` | An optional but runnable extension notebook | Use it to go deeper after the core path is stable |
| `note-first` | A literature-first notebook that frames a frontier area | Convert only after the repo has enough bandwidth for serious implementation |

At the moment, most notebooks are still `scaffold`. That is acceptable for a v1 restructure. What matters is that the repo now has an honest map and a disciplined plan for expansion.

````python
def list_notebooks(folder: Path):
    return sorted(path.relative_to(root) for path in folder.rglob("*.ipynb"))


for lane in ["00_Lab_Guide", "01_Core_Path", "10_Research_Tracks", "20_Frontier_Notes"]:
    print(f"\n{lane}")
    print("-" * len(lane))
    for path in list_notebooks(root / lane):
        print(path)
````

## Scope Discipline

The restructure is only useful if future notebook authoring stays disciplined.

### Core path rules

- one method family
- one canonical PDE or dynamical system
- one main training loop
- one baseline comparison
- one or two ablations at most
- one explicit failure mode worth understanding

### Research-track rules

- broader or more specialized than the core path
- allowed to assume the reader already finished the core path
- can branch into a subtopic family, but should still avoid becoming a whole survey paper in one notebook

### Frontier-note rules

Every frontier notebook should answer four questions before any heavy implementation begins:

1. Why is this topic worth caring about?
2. Which papers are the best starting points?
3. What toy feasibility check would be reasonable next?
4. What is intentionally deferred for later?

## Experiment Protocol

Every future executable notebook should make the experiment design obvious near the top. At minimum, record:

1. **Research question**
2. **Canonical problem**
3. **Model family**
4. **Data and physics supervision setup**
5. **Baseline**
6. **Metrics**
7. **Runtime budget**
8. **Known failure mode to watch**
9. **References**
10. **Next notebook(s)**

Recommended habits for this repo:

- Start with the smallest defensible problem before adding complexity.
- Fix seeds whenever stochasticity matters.
- Compare against at least one simpler baseline whenever possible.
- Separate observations from claims. A pretty plot is not a conclusion.
- Make failure analysis part of the notebook, not an afterthought.
- If a notebook starts covering multiple paper families or multiple major geometries, split it.
